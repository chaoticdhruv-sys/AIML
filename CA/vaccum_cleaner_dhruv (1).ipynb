Smart Office Cleaning Agent Using Artificial Intelligence

**Aim:**
To build a small AI-based cleaning agent for an office environment. The agent should look at the condition of different areas, decide which area needs attention first, move there, clean it, and keep track of its battery while working.

---

**Experimental Objective:**

* Understand how a simple intelligent agent observes an environment and takes action.
* Represent different office areas and their conditions using Python.
* Give the cleaning agent a basic decision-making strategy instead of making it clean areas in a fixed order.
* Study how battery level and room availability can affect the agent's decisions.

---

**Expected Outcome:**

* Understand the basic idea of an AI agent.
* Create an office environment using Python data structures and classes.
* Allow the agent to choose an area based on its current condition.
* Observe how the agent changes the environment after cleaning.
* Track the agent's battery and performance during the cleaning process.

---

**Concept:**

A simple AI agent observes what is happening around it and uses that information to decide what to do next. It does not need a complicated learning model for this experiment; a few sensible rules are enough to demonstrate intelligent behaviour.

For this practical:

**Environment → Office**

1. **Agent →** Smart Vacuum Cleaner
2. **Goal →** Keep accessible office areas clean
3. **Perception →** Dirt level, priority, and occupancy of each area
4. **Actions →** Select an area, move there, clean it, or recharge
5. **Performance →** Cleaning progress while avoiding unnecessary battery usage


**Flow Chart**

                   ↓
          Set Up Office Areas
                   ↓
           Start Cleaning Agent
                   ↓
          Read Area Conditions
                   ↓
       Is Battery Enough to Work?
             ┌─────┴─────┐
            NO           YES
             │            │
             ↓            │
          Recharge         │
             │            │
             └─────┬──────┘
                   ↓
       Find Accessible Dirty Areas
                   ↓
       Give Each Area a Need Score
                   ↓
       Pick the Highest-Need Area
                   ↓
             Move and Clean
                   ↓
        Update Dirt, Battery & Score
                   ↓
             More Areas to Clean?
             │            │
            YES           NO
             │            │
             └─────→     END



```python
import random

class OfficeEnvironment:

    def __init__(self):
        self.areas = {
            "Manager Cabin": {
                "dirt": random.randint(25, 85),
                "priority": 4,
                "occupied": random.choice([True, False]),
                "distance": 2
            },

            "Conference Room": {
                "dirt": random.randint(25, 85),
                "priority": 3,
                "occupied": random.choice([True, False]),
                "distance": 3
            },

            "Work Area": {
                "dirt": random.randint(25, 85),
                "priority": 5,
                "occupied": random.choice([True, False]),
                "distance": 2
            },

            "Server Room": {
                "dirt": random.randint(25, 85),
                "priority": 5,
                "occupied": random.choice([True, False]),
                "distance": 4
            },

            "Reception Area": {
                "dirt": random.randint(25, 85),
                "priority": 2,
                "occupied": random.choice([True, False]),
                "distance": 1
            },

            "Main Corridor": {
                "dirt": random.randint(25, 85),
                "priority": 3,
                "occupied": False,
                "distance": 1
            }
        }


class SmartVacuumAgent:

    def __init__(self, office):
        self.office = office
        self.location = "Main Corridor"
        self.battery = 100
        self.score = 0
        self.cleaned_areas = 0

    def show_environment(self):
        print("\nCurrent Office Condition")
        print("------------------------")

        for area, info in self.office.areas.items():
            print(
                area,
                "| Dirt:", info["dirt"],
                "| Priority:", info["priority"],
                "| Occupied:", info["occupied"]
            )

    def need_score(self, area):
        info = self.office.areas[area]

        # Occupied rooms are temporarily ignored.
        if info["occupied"] or info["dirt"] <= 10:
            return -1

        # Dirt matters most, while priority and distance influence the choice.
        return (info["dirt"] * 1.5) + (info["priority"] * 12) - (info["distance"] * 4)

    def choose_area(self):
        candidates = {}

        for area, info in self.office.areas.items():
            score = self.need_score(area)

            if score >= 0:
                candidates[area] = score

        if not candidates:
            return None

        return max(candidates, key=candidates.get)

    def move_to(self, area):
        info = self.office.areas[area]
        movement_cost = info["distance"] * 2

        print("\nMoving from", self.location, "to", area)

        self.location = area
        self.battery -= movement_cost
        self.score -= movement_cost // 2

    def clean_area(self, area):
        info = self.office.areas[area]

        if info["occupied"]:
            print(area, "is currently occupied. Cleaning postponed.")
            return False

        if info["dirt"] <= 10:
            print(area, "does not need further cleaning.")
            return False

        print("Cleaning", area)
        print("Dirt before cleaning:", info["dirt"])

        # The agent removes most, but not necessarily all, of the dirt.
        removed = max(10, int(info["dirt"] * 0.75))
        info["dirt"] = max(0, info["dirt"] - removed)

        cleaning_cost = 12
        self.battery -= cleaning_cost
        self.score += removed // 4 + info["priority"]
        self.cleaned_areas += 1

        print("Dirt removed:", removed)
        print("Dirt remaining:", info["dirt"])

        return True

    def recharge(self):
        print("\nBattery is getting low.")
        print("Returning to charging point...")

        self.location = "Charging Station"
        self.battery = 100

        print("Battery is now 100%.")

    def run_cleaning_cycle(self):
        print("\nSmart vacuum starts its cleaning cycle")
        print("---------------------------------------")

        while True:
            area = self.choose_area()

            if area is None:
                print("\nNo accessible area currently needs cleaning.")
                break

            required_battery = (self.office.areas[area]["distance"] * 2) + 12

            if self.battery < required_battery:
                self.recharge()

            self.move_to(area)
            cleaned = self.clean_area(area)

            if not cleaned:
                # Avoid repeatedly selecting an area that became unavailable.
                self.office.areas[area]["dirt"] = 0

            print("Battery:", self.battery)
            print("Score:", self.score)

    def final_report(self):
        print("\nFinal Office Report")
        print("-------------------")

        for area, info in self.office.areas.items():
            print(area, "| Remaining Dirt:", info["dirt"])

        print("\nAgent Location:", self.location)
        print("Battery Left:", self.battery)
        print("Areas Visited for Cleaning:", self.cleaned_areas)
        print("Final Performance Score:", self.score)


# Create the office environment
office = OfficeEnvironment()

# Create the intelligent cleaning agent
vacuum = SmartVacuumAgent(office)

# Display the starting condition
vacuum.show_environment()

# Run the decision-making cleaning cycle
vacuum.run_cleaning_cycle()

# Display the final result
vacuum.final_report()
```

**Conclusion:** The smart office vacuum agent was implemented successfully using Python and simple AI decision-making rules.

Instead of visiting every area in a fixed sequence, the agent looks at the current condition of the office and gives each accessible area a need score. Dirt level and area importance increase the score, while distance slightly reduces it. This makes the agent choose a location based on the situation at that moment.

The cleaning process also considers battery usage. When there is not enough battery for the next task, the agent recharges before continuing. Cleaning reduces the dirt level gradually, so the final state can be different from the starting state in a more realistic way.

Overall, the practical shows how even a small set of rules can make a program behave like an intelligent agent that observes, decides, acts, and updates its environment.
