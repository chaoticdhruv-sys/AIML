**Theory**<br>
Prolog is a logic programming language used in Artificial Intelligence and knowledge-based systems.
A Prolog program consists mainly of facts, rules, and queries.
Facts represent known information, while rules define relationships between facts.
Queries are used to retrieve information and verify conditions from the knowledge base.
In this experiment, Prolog is used to represent and retrieve information about travel destinations, destination types, weather, budgets, and activities.
<br><br>
**Algorithm**<br>
Start the program.
Define the travel destinations using the destination predicate.
Define destination types using the type predicate.
Define weather conditions using the weather predicate.
Define budget categories using the budget predicate.
Define available activities using the activity predicate.
Define rules for identifying:
Beach destinations
Mountain destinations
Budget-friendly destinations
Medium-budget destinations
Adventure destinations
Sightseeing destinations

Cold-weather destinations
Execute different queries.
Prolog matches the queries with facts and rules.
Display the results.
Stop the program.

```python
% Facts

% Destinations
destination(goa).
destination(manali).
destination(jaipur).
destination(mumbai).
destination(kerala).
destination(srinagar).

% Destination type
type(goa,beach).
type(manali,mountain).
type(jaipur,historical).
type(mumbai,city).
type(kerala,nature).
type(srinagar,mountain).

% Weather
weather(goa,sunny).
weather(manali,cold).
weather(jaipur,hot).
weather(mumbai,humid).
weather(kerala,rainy).
weather(srinagar,cold).

% Budget category
budget(goa,medium).
budget(manali,medium).
budget(jaipur,low).
budget(mumbai,high).
budget(kerala,medium).
budget(srinagar,high).

% Activities
activity(goa,water_sports).
activity(goa,nightlife).
activity(manali,trekking).
activity(manali,camping).
activity(jaipur,sightseeing).
activity(jaipur,fort_visit).
activity(mumbai,shopping).
activity(mumbai,sightseeing).
activity(kerala,boating).
activity(kerala,nature_walk).
activity(srinagar,sightseeing).
activity(srinagar,boating).

% Rules

recommend_beach(Destination) :-
    destination(Destination),
    type(Destination,beach).

recommend_mountain(Destination) :-
    destination(Destination),
    type(Destination,mountain).

budget_trip(Destination) :-
    destination(Destination),
    budget(Destination,low).

medium_budget_trip(Destination) :-
    destination(Destination),
    budget(Destination,medium).

adventure_trip(Destination) :-
    activity(Destination,trekking).

adventure_trip(Destination) :-
    activity(Destination,water_sports).

sightseeing_trip(Destination) :-
    activity(Destination,sightseeing).

nature_trip(Destination) :-
    type(Destination,nature).

cold_destination(Destination) :-
    destination(Destination),
    weather(Destination,cold).

recommend(Destination,beach) :-
    type(Destination,beach).

recommend(Destination,mountain) :-
    type(Destination,mountain).

recommend(Destination,historical) :-
    type(Destination,historical).

recommend(Destination,nature) :-
    type(Destination,nature).
```
